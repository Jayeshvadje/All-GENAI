```python
def factorial(n):
    """This function gives the factorial of the number"""
    factorial = 1
    for i in range(1,n+1):
        factorial = factorial * i
    return factorial
factorial(5)
```

```python
from langchain_core.runnables import RunnableParallel,RunnableLambda
def upper(data):
    return data.upper()
lam = RunnableLambda(upper)
def lower(data):
    return data.lower()
lam2 =RunnableLambda(lower)

parallel = RunnableParallel(upper =lam,
                            lower = lam2)
print(parallel.invoke("hello woRld"))
```

```python
from langchain_core.runnables import RunnableLambda,RunnableParallel,RunnablePassthrough

word = str(input("Enter the input"))

def upper(word):
    return word.upper()
lam = RunnableLambda(upper)

def len_word(word):
    return len(word)
lam1 = RunnableLambda(len_word)

parallel = RunnableParallel(original = RunnablePassthrough(),
                            upper = lam,
                            len = lam1)
result = parallel.invoke(word)
print(result)
```

```python
from langchain_core.runnables import RunnableLambda,RunnableSequence

def double(x):
    return x*2

def add_ten(x):
    return x+10

double_lambda = RunnableLambda(double)

add_ten_lambda = RunnableLambda(add_ten)

Sequence = RunnableSequence(double_lambda,
                            add_ten_lambda)

result = Sequence.invoke(5)
print(result)
```

```python
from dotenv import load_dotenv

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

load_dotenv()


# -----------------------------
# 1. Create the model
# -----------------------------

model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)


# -----------------------------
# 2. Create the prompt
# -----------------------------

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful AI assistant. "
        "Use the conversation history to answer the user's questions."
    ),

    # Previous conversation will be inserted here
    MessagesPlaceholder(variable_name="history"),

    # Current user question
    ("human", "{question}")
])


# -----------------------------
# 3. Create normal chain
# -----------------------------

chain = prompt | model


# -----------------------------
# 4. Create memory store
# -----------------------------

store = {}


def get_session_history(session_id):

    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()

    return store[session_id]


# -----------------------------
# 5. Add conversation memory
# -----------------------------

conversation = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="question",
    history_messages_key="history"
)


# -----------------------------
# 6. First question
# -----------------------------

result = conversation.invoke(
    {
        "question": "My name is Priyanka and I am learning Python."
    },
    config={
        "configurable": {
            "session_id": "user_1"
        }
    }
)

print("AI:", result.content)


# -----------------------------
# 7. Second question
# -----------------------------

result = conversation.invoke(
    {
        "question": "What am I learning?"
    },
    config={
        "configurable": {
            "session_id": "user_1"
        }
    }
)

print("AI:", result.content)
```
