```python
from langchain_core.runnables import RunnableLambda,RunnableParallel,RunnablePassthrough

word = str(input("Enter the input"))

def upper(word):
    return word.upper()
lam = RunnableLambda(upper)

def len_word(word):
    return len(word)
lam1 = RunnableLambda(len_word)

parallel = RunnableParallel(original = RunnablePassthrough(),
                            upper = lam,
                            len = lam1)
result = parallel.invoke(word)
print(result)
```

```python
from dotenv import load_dotenv

from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
from langchain_core.chat_history import InMemoryChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory

load_dotenv()


# -----------------------------
# 1. Create the model
# -----------------------------

model = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash"
)


# -----------------------------
# 2. Create the prompt
# -----------------------------

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are a helpful AI assistant. "
        "Use the conversation history to answer the user's questions."
    ),

    # Previous conversation will be inserted here
    MessagesPlaceholder(variable_name="history"),

    # Current user question
    ("human", "{question}")
])


# -----------------------------
# 3. Create normal chain
# -----------------------------

chain = prompt | model


# -----------------------------
# 4. Create memory store
# -----------------------------

store = {}


def get_session_history(session_id):

    if session_id not in store:
        store[session_id] = InMemoryChatMessageHistory()

    return store[session_id]


# -----------------------------
# 5. Add conversation memory
# -----------------------------

conversation = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="question",
    history_messages_key="history"
)


# -----------------------------
# 6. First question
# -----------------------------

result = conversation.invoke(
    {
        "question": "My name is Priyanka and I am learning Python."
    },
    config={
        "configurable": {
            "session_id": "user_1"
        }
    }
)

print("AI:", result.content)


# -----------------------------
# 7. Second question
# -----------------------------

result = conversation.invoke(
    {
        "question": "What am I learning?"
    },
    config={
        "configurable": {
            "session_id": "user_1"
        }
    }
)

print("AI:", result.content)
```

```python
from langchain_core.tools import tool
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
load_dotenv()
@tool
def multiply(a:float, b:float) ->float:
    """multiply two numbers together"""
    return a*b

model  = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash")
@tool
def addition(a:float, b:float) ->float:
    """add two numbers together"""
    return a+b

model_with_tool= model.bind_tools([addition,multiply])
result = model_with_tool.invoke("what is addition and multiplication of 25 and 6?")
print(result.tool_calls)


tool_call = result.tool_calls[0]
answer = multiply.invoke(tool_call["args"])
answer2 = addition.invoke(tool_call["args"])
print(answer,answer2)
```

```python
from langchain_core.tools import tool
from langchain_google_genai import ChatGoogleGenerativeAI
import os
from dotenv import load_dotenv
load_dotenv()
import requests

@tool
def get_whether(city:str) ->str:
    """Get the current whether of the city"""
    api_key = os.getenv("OPENWEATHER_API_KEY")
    url = "https://api.openweathermap.org/data/2.5/weather"
    params = {"q": city, "appid": api_key, "units": "metric"}
    response = requests.get(url, params)
    # data = response.json()
    # temperature = data["main"]["temp"]
    # description = data["weather"][0]["description"]
    #
    # return f"The temperature in {city} is {temperature}°C with {description}."
    response = requests.get(url, params=params)

    print("Status code:", response.status_code)
    print("Response:", response.json())

    return "test"

```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash")
model_with_tool = model.bind_tools([get_whether])
result = model_with_tool.invoke("what is the whether in pune?")
answer = result.tool_calls[0]
final_answer = get_whether.invoke(answer["args"])
print(final_answer)
```

```python
from langchain_core.messages import ToolMessage
tool_message = ToolMessage(content= str(final_answer),
                           tool_call_id = answer["id"])
messages = [
    ("human", result),
    result,
    tool_message
]
final_answer = model_with_tool.invoke(messages)
print(final_answer)
```

```python
model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash"
)

model_with_tool = model.bind_tools([get_whether])

# 1. User question
question = "What is the weather in Pune?"

# 2. Gemini decides to call the tool
result = model_with_tool.invoke(question)

# 3. Get the tool call
answer = result.tool_calls[0]

# 4. Execute the tool
tool_result = get_whether.invoke(answer["args"])

print("Tool result:", tool_result)

# 5. Create ToolMessage
from langchain_core.messages import ToolMessage

tool_message = ToolMessage(
    content=str(tool_result),
    tool_call_id=answer["id"]
)

# 6. Send conversation + tool result back to Gemini
messages = [
    ("human", question),
    result,
    tool_message
]

# 7. Gemini creates the final answer
final_result = model_with_tool.invoke(messages)

print(final_result.content)
```
